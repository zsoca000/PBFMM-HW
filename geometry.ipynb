```python
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import numpy as np

def draw_rectangle(ax,x,y,w,h,dx,facecolor):
    X,Y = np.meshgrid(
        np.arange(x + dx/2, x + w - dx/2 + 1e-9, dx),
        np.arange(y + dx/2, y + h - dx/2 + 1e-9, dx)
    )
    
    print(X.size)
    square = patches.Rectangle((x, y), w, h, edgecolor='none',facecolor=facecolor)
    ax.add_patch(square)
    ax.plot(X, Y, '.', color='black', markersize=1)
    
    return X.size

L = 0.2

x_c, y_c = 0.5*L, 1.25*L

csize =  0.005
dx = csize/2.5
h = csize/2
```

```python
L_obj = 0.025

fig, ax = plt.subplots(figsize=(12,8),dpi=150)

num_nodes = 0  
draw_rectangle(ax,0,0,L,L,dx,facecolor='lightgreen')

x,y = x_c - L_obj/2, y_c - L_obj/2
w,h = L_obj, L_obj

num_nodes += draw_rectangle(ax,x,y,w,h,dx,facecolor='red')

ax.set_xlim(0, L)
ax.set_ylim(0, 1.5*L)
ax.set_aspect('equal')

print(f'# of nodes: {num_nodes}')
plt.grid(True)
plt.show()
```

```python
t = 0.01
W = 0.05
H = 0.025

fig, ax = plt.subplots(figsize=(12,8),dpi=150)
num_nodes = 0

draw_rectangle(ax,0,0,L,L,dx,facecolor='lightgreen')


(x,y) = (x_c - W/2, y_c - H/2)
(w,h) = (t,H)
num_nodes += draw_rectangle(ax,x,y,w,h,dx,facecolor='red')

(x,y) = (x_c - W/2+t, y_c - H/2)
(w,h) = (W-2*t,t)
num_nodes += draw_rectangle(ax,x,y,w,h,dx,facecolor='blue')

(x,y) = (x_c + W/2 - t, y_c - H/2)
(w,h) = (t,H)
num_nodes +=draw_rectangle(ax,x,y,w,h,dx,facecolor='red')

ax.set_xlim(0, L)
ax.set_ylim(0, 1.5*L)
ax.set_aspect('equal')

print(f'# of nodes: {num_nodes}')
with open(f"boat_{t}.txt", "w") as f:
    f.write(" ".join(str(i) for i in range(num_nodes)))
plt.grid(True)
plt.show()
```

```python
t = 0.0075
W = 0.075
H = 0.025

L_obj = 0.025


fig, ax = plt.subplots(figsize=(8,20),dpi=150)
num_nodes_load = 0
num_nodes_boat = 0
num_nodes_water = 0
num_nodes_water += draw_rectangle(ax,0,0,L,L,dx,facecolor='lightgreen')

(x,y) = (x_c - L_obj/2, 2.25*L-L_obj/2)
(w,h) = (L_obj,L_obj)
num_nodes_load += draw_rectangle(ax,x,y,w,h,dx,facecolor='red')

(x,y) = (x_c - W/2, L)
(w,h) = (t,H)
num_nodes_boat += draw_rectangle(ax,x,y,w,h,dx,facecolor='red')

(x,y) = (x_c - W/2+t, L)
(w,h) = (W-2*t,t)
num_nodes_boat += draw_rectangle(ax,x,y,w,h,dx,facecolor='blue')

(x,y) = (x_c + W/2 - t, L)
(w,h) = (t,H)
num_nodes_boat +=draw_rectangle(ax,x,y,w,h,dx,facecolor='red')

ax.set_xlim(0, L)
ax.set_ylim(0, 2.5*L)
ax.set_aspect('equal')

print(f'# of nodes water: {num_nodes_water}')
print(f'# of nodes boat: {num_nodes_boat}')
print(f'# of nodes load: {num_nodes_load}')
print(f'# of nodes SUM: {num_nodes_water+num_nodes_boat+num_nodes_load}')
with open(f"boat.txt", "w") as f:
    f.write(" ".join(str(i) for i in range(num_nodes_load)))
    f.write('\n')
    f.write(" ".join(str(i) for i in range(num_nodes_load,num_nodes_load+num_nodes_boat)))
plt.grid(True)
plt.show()
```
